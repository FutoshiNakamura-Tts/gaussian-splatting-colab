このノートブックは、Colabのプロキシ機能(`proxyPort`)に依存せず、データを表示するさまざまな方法を検証します。

**検証項目:**
1.  **静止画 (メモリ)**: `matplotlib` / `PIL` によるインメモリ表示。
2.  **静止画 (ファイル)**: 生成した画像ファイルを Markdown 形式 `![alt](path)` で表示できるか。**[結果: 失敗]**
3.  **静止画 (Python埋め込み)**: `IPython.display.Image` を使用して表示できるか。**[追加検証]**
4.  **インタラクティブグラフ**: `plotly` を使用した3D表示。
5.  **ビデオ**: MP4動画の埋め込み。

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display, Image as IPImage, Markdown
import plotly.graph_objects as go
import os

print("Libraries imported.")
```

```python
# 1. 静止画の生成と表示 (Matplotlib & PIL)
print("=== 1. Static Image Test (Memory) ===")

# ランダムノイズ画像の生成
im_data = np.random.rand(100, 100, 3)

plt.figure(figsize=(4, 4))
plt.imshow(im_data)
plt.title("Matplotlib Display (In-Memory)")
plt.axis('off')
plt.show()

print("Matplotlib rendering complete.")
```

```python
# 2. Markdownによるローカルファイル表示
print("=== 2. Local File Markdown Test ===")

# テスト用画像の保存
filename = "test_image.png"
plt.figure(figsize=(4, 4))
plt.imshow(im_data)
plt.title("Saved Image File")
plt.axis('off')
plt.savefig(filename)
plt.close()
print(f"Saved {filename}.")

# パス検証
abs_path = os.path.abspath(filename)
print(f"Absolute Path: {abs_path}")
print(f"Current Working Directory: {os.getcwd()}")

if os.path.exists(abs_path):
    print("[✅] File successfully created at destination.")
else:
    print("[❌] File NOT found at destination!")

# Markdownとして表示を試みる（失敗想定）
md_text = f"""
### Markdown Display Test

**Test A: Relative Path** (`{filename}`) - Likely Broken
![Relative Path Image]({filename})

**Test B: Absolute Path** (`{abs_path}`) - Likely Broken
![Absolute Path Image]({abs_path})
"""

display(Markdown(md_text))

print("Markdown display cell executed.")
```

```python
# 3. Python関数による画像ファイル表示
print("=== 3. Python Display Function Test ===")

if os.path.exists(filename):
    print(f"Displaying {filename} using IPython.display.Image...")
    display(IPImage(filename=filename))
    print("[✅] Image displayed successfully (using data embedding).")
else:
    print("Test file not found.")
```

```python
# 4. インタラクティブ3Dグラフ (Plotly)
print("=== 4. Interactive 3D Plot (Plotly) ===")
print("This uses JavaScript injection directly into the cell output, bypassing proxyPort.")

# ランダムな3D点群データ
count = 500
x = np.random.randn(count)
y = np.random.randn(count)
z = np.random.randn(count)

fig = go.Figure(data=[go.Scatter3d(
    x=x, y=y, z=z,
    mode='markers',
    marker=dict(
        size=5,
        color=z,
        colorscale='Viridis',
        opacity=0.8
    )
)])

fig.update_layout(margin=dict(l=0, r=0, b=0, t=0), height=500, title="3D Scatter Plot")
fig.show()

print("Plotly rendering command sent.")
```
