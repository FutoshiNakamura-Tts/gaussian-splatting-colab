# Debug Wheels Notebook
Verifies the installation of `diff-gaussian-rasterization` and `simple-knn` wheels.


```python
import os
import subprocess
import sys

def run_command(cmd):
    print(f"Executing: {cmd}")
    subprocess.run(cmd, shell=True, check=False)

print(f"Python Version: {sys.version}")

# 1. Clone Wheels Repo (Simulating Colab environment)
wheels_repo_url = "https://github.com/FutoshiNakamura-Tts/gaussian-splatting-colab"
branch = "colab-t4-2025-12-18"
target_dir = "/content/wheels_repo"

if not os.path.exists(target_dir):
    print("Cloning repo...")
    run_command(f"git clone -b {branch} {wheels_repo_url} {target_dir}")
else:
    print("Repo already exists.")

wheels_dir = f"{target_dir}/wheels"
print(f"Checking wheels in {wheels_dir}...")
run_command(f"ls -R {wheels_dir}")
```

```python
# 2. Attempt Installation
print("Installing wheels...")
# Find wheels
import glob
wheels = glob.glob(f"{wheels_dir}/*.whl")

if wheels:
    for w in wheels:
        run_command(f"pip install {w}")
else:
    print("No wheels found in root wheels dir.")

print("Checking 3dgs_accel...")
accel_wheels = glob.glob(f"{wheels_dir}/3dgs_accel/*.whl")
if accel_wheels:
    for w in accel_wheels:
         run_command(f"pip install {w}")
```

```python
# 3. Verify Import
try:
    import diff_gaussian_rasterization
    print("SUCCESS: diff_gaussian_rasterization imported.")
except ImportError as e:
    print(f"FAIL: {e}")

try:
    import simple_knn
    print("SUCCESS: simple_knn imported.")
except ImportError as e:
    print(f"FAIL: {e}")
```
